# Exploratory Data Analysis (EDA)
# M5 Forecasting Dataset

This notebook explores the M5 dataset to understand:
- Data structure and quality
- Demand patterns and trends
- Seasonality (daily, weekly, monthly)
- Product/store characteristics
- Relationships between features

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data.preprocessing import load_processed_data
from src.utils.logger import setup_logger

logger = setup_logger(__name__)

# Configure plotting
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline
```

## 1. Load Processed Data

```python
# Load preprocessed data
train_df, val_df, test_df = load_processed_data('../data/processed')

print(f"Train shape: {train_df.shape}")
print(f"Val shape: {val_df.shape}")
print(f"Test shape: {test_df.shape}")
```

## 2. Dataset Overview

```python
# Display sample data
train_df.head()
```

```python
# Data types and missing values
train_df.info()
```

## 3. Demand Analysis

```python
# Overall demand statistics
print("Sales Statistics:")
print(train_df['sales'].describe())

# Zero demand analysis
zero_sales_pct = (train_df['sales'] == 0).mean() * 100
print(f"\nZero sales percentage: {zero_sales_pct:.2f}%")
```

```python
# Sales distribution
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Histogram
axes[0].hist(train_df['sales'], bins=50, edgecolor='black')
axes[0].set_xlabel('Sales')
axes[0].set_ylabel('Frequency')
axes[0].set_title('Sales Distribution')

# Box plot
axes[1].boxplot(train_df['sales'])
axes[1].set_ylabel('Sales')
axes[1].set_title('Sales Box Plot')

plt.tight_layout()
plt.savefig('../reports/figures/sales_distribution.png', dpi=300)
plt.show()
```

## 4. Time Series Patterns

```python
# Aggregate daily sales
daily_sales = train_df.groupby('date')['sales'].sum().reset_index()

plt.figure(figsize=(15, 6))
plt.plot(daily_sales['date'], daily_sales['sales'])
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.title('Daily Aggregate Sales Over Time')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('../reports/figures/daily_sales_trend.png', dpi=300)
plt.show()
```

## 5. Store Analysis

```python
# Sales by store
store_sales = train_df.groupby('store_id')['sales'].agg(['sum', 'mean', 'std']).reset_index()
store_sales.sort_values('sum', ascending=False, inplace=True)

print("Top 5 Stores by Total Sales:")
print(store_sales.head())
```

## 6. Category Analysis

```python
# Sales by category
category_sales = train_df.groupby('cat_id')['sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
plt.bar(category_sales['cat_id'], category_sales['sales'])
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.title('Sales by Category')
plt.tight_layout()
plt.savefig('../reports/figures/category_sales.png', dpi=300)
plt.show()
```

## 7. Price Analysis

```python
# Price distribution
print("Price Statistics:")
print(train_df['sell_price'].describe())

# Price vs Sales correlation
correlation = train_df[['sell_price', 'sales']].corr()
print(f"\nPrice-Sales Correlation: {correlation.iloc[0, 1]:.3f}")
```

## 8. Seasonality Patterns

```python
# Day of week analysis
train_df['day_of_week'] = train_df['date'].dt.dayofweek
dow_sales = train_df.groupby('day_of_week')['sales'].mean().reset_index()

day_names = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
dow_sales['day_name'] = dow_sales['day_of_week'].map(lambda x: day_names[x])

plt.figure(figsize=(10, 6))
plt.bar(dow_sales['day_name'], dow_sales['sales'])
plt.xlabel('Day of Week')
plt.ylabel('Average Sales')
plt.title('Weekly Seasonality - Average Sales by Day of Week')
plt.tight_layout()
plt.savefig('../reports/figures/weekly_seasonality.png', dpi=300)
plt.show()
```
